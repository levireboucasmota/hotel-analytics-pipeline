```python
import pandas as pd

df = pd.read_csv("../data/raw/hotel_bookings.csv")
df_limpo = df.copy()
print(f"Linhas iniciais: {len(df_limpo):,}")
```

```python
antes = len(df_limpo)
df_limpo = df_limpo.drop_duplicates()
print(f"Duplicatas removidas: {antes - len(df_limpo):,}")
```

```python
# adr negativo
antes = len(df_limpo)
df_limpo = df_limpo[df_limpo["adr"] >= 0]
print(f"adr negativo removido: {antes - len(df_limpo)}")

# outlier de 5400
antes = len(df_limpo)
df_limpo = df_limpo[df_limpo["adr"] < 5000]
print(f"Outlier adr removido: {antes - len(df_limpo)}")

# reservas sem hóspedes
antes = len(df_limpo)
total_hospedes = df_limpo["adults"] + df_limpo["children"].fillna(0) + df_limpo["babies"]
df_limpo = df_limpo[total_hospedes > 0]
print(f"Sem hóspedes removidas: {antes - len(df_limpo)}")
```

```python
df_limpo["children"] = df_limpo["children"].fillna(0).astype(int)
df_limpo["country"] = df_limpo["country"].fillna("Desconhecido")
df_limpo["agent"] = df_limpo["agent"].fillna(0).astype(int)
df_limpo["meal"] = df_limpo["meal"].replace("Undefined", "SC")
```

```python
df_limpo["tem_empresa"] = df_limpo["company"].notna().astype(int)
df_limpo = df_limpo.drop(columns=["company"])
```

```python
df_limpo["adr_zero"] = (df_limpo["adr"] == 0).astype(int)
df_limpo["sem_noites"] = (
    (df_limpo["stays_in_weekend_nights"] + df_limpo["stays_in_week_nights"]) == 0
).astype(int)
```

```python
print(f"Linhas finais: {len(df_limpo):,}")
print(f"Taxa de cancelamento antes:  {df['is_canceled'].mean():.4f}")
print(f"Taxa de cancelamento depois: {df_limpo['is_canceled'].mean():.4f}")

print(df_limpo.isna().sum()[df_limpo.isna().sum() > 0])   # deve ficar vazio
print((df_limpo["adr"] < 0).sum())                          # deve ser 0
print(df_limpo["adr"].max())                                # deve ser 510
```

```python
df_limpo.to_csv("../data/staging/hotel_bookings_limpo.csv", index=False)
```
